````python
# Función que recibe un texto y cuenta
# cuántas letras mayúsculas y minúsculas contiene

text = "Kim Seokjin is mr. Worldwide Handsome"

def count_case(text):

    # Creamos dos contadores.
    # Ambos empiezan en 0 porque todavía no hemos revisado ninguna letra.
    upper = 0
    lower = 0

    # Recorremos el texto carácter por carácter.
    # No necesitamos convertir el string en una lista:
    # los strings también se pueden recorrer con un for.
    for character in text:

        # .isupper() comprueba si el carácter es una letra mayúscula.
        # Devuelve True o False.
        if character.isupper():

            # Si encontramos una mayúscula,
            # aumentamos el contador upper en 1.
            # Es lo mismo que: upper = upper + 1
            upper += 1

        # Si no era mayúscula, comprobamos si es minúscula.
        # Usamos elif y no else porque un carácter también
        # podría ser un espacio, número, signo de puntuación, etc.
        elif character.islower():

            # Si encontramos una minúscula,
            # aumentamos el contador lower en 1.
            lower += 1

    # Cuando el for termina de recorrer todo el texto,
    # devolvemos los dos resultados.
    # Al devolver dos valores separados por una coma,
    # Python los devuelve juntos en una tupla.
    return upper, lower


# Probamos la función con un texto.
result = count_case("Hello World")

# Mostramos el resultado en pantalla.
print(result)

````

````python
# Importamos el módulo string.
# Lo necesitamos porque contiene string.punctuation,
# que incluye los signos de puntuación habituales.
import string


# Creamos una función que recibe una frase como parámetro.
def remove_punctuation(sentence):

    # Creamos una lista vacía donde iremos guardando
    # los caracteres que NO sean signos de puntuación.
    clean_sentence = []

    # Recorremos la frase carácter por carácter.
    # Un string se puede recorrer directamente con un for.
    for character in sentence:

        # Comprobamos si el carácter actual NO está
        # dentro de los signos de puntuación.
        if character not in string.punctuation:

            # Si no es puntuación, lo añadimos
            # a nuestra lista de caracteres.
            clean_sentence.append(character)

    # clean_sentence es una lista de caracteres.
    # "".join(...) une todos esos caracteres sin añadir
    # nada entre ellos y los convierte en un único string.
    #
    # Ejemplo:
    # ["H", "e", "l", "l", "o"] → "Hello"
    #
    # return devuelve ese resultado fuera de la función.
    return "".join(clean_sentence)


# Probamos nuestra función.
result = remove_punctuation("Hello!!! How are you?")

# Mostramos el resultado.
print(result)
````

There are several ways to import functions from a Python module such as functions.py to a Jupyter Notebook:

1. Importing specific functions: If you only need to use a few functions from the module, you can import them individually using the from keyword. This way, you can call the functions directly using their names, without having to use the module name. For example:

```python
from function_file import function_name

function_name(arguments)```

2. Importing the entire module: You can import the entire module using the import keyword followed by the name of the module. Then, you can call the functions using the module_name.function_name() syntax. Example:

```python
import function_file

function_file.function_name()
```

3. Renaming functions during import: You can also rename functions during import using the as keyword. This is useful if you want to use a shorter or more descriptive name for the function in your code. For example:

```python
from function_file import function_name as f

f.function_name(arguments)
```

Regardless of which method you choose, make sure that the functions.py file is in the same directory as your Jupyter Notebook, or else specify the path to the file in the import statement or by using `sys.path`.
